# 02 - Visualisation des données

## Objectifs
- Analyser visuellement les données
- Comprendre les relations entre les variables
- Identifier les variables pertinentes pour la régression


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
DATA_PATH = "../data/raw/housing.csv"
df = pd.read_csv(DATA_PATH)
```

```python
plt.figure()
plt.hist(df["price"], bins=30)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution des prix")
plt.show()
```

```python
plt.figure()
plt.boxplot(df["price"])
plt.ylabel("Price")
plt.title("Boxplot des prix")
plt.show()
```

```python
plt.figure()
plt.scatter(df["area"], df["price"])
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Surface vs Prix")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Matrice de corrélation")
plt.show()
```

```python
cat_col = "furnishingstatus"  # adapte si besoin

plt.figure()
sns.boxplot(x=df[cat_col], y=df["price"])
plt.xticks(rotation=45)
plt.title(f"{cat_col} vs Prix")
plt.show()
```
